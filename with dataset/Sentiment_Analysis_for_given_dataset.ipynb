```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("/content/drive/MyDrive/Datasets/Tweets.csv")

# -----------------------------
# 1. Overall Sentiment Count
# -----------------------------
sentiment_counts = df["airline_sentiment"].value_counts()

print("\nOverall Sentiment Distribution:")
print(sentiment_counts)

# Plot
plt.figure()
sentiment_counts.plot(kind='bar')
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Tweets")
plt.show()

# -----------------------------
# 2. Positive Tweets Analysis
# -----------------------------
positive_tweets = df[df["airline_sentiment"] == "positive"]

print("\nTotal Positive Tweets:", len(positive_tweets))
print("\nSample Positive Tweets:")
print(positive_tweets["text"].head(5))

# -----------------------------
# 3. Negative Tweets Analysis
# -----------------------------
negative_tweets = df[df["airline_sentiment"] == "negative"]

print("\nTotal Negative Tweets:", len(negative_tweets))
print("\nSample Negative Tweets:")
print(negative_tweets["text"].head(5))

# -----------------------------
# 4. Negative Reason Analysis
# -----------------------------
neg_reason = negative_tweets["negativereason"].value_counts()

print("\nTop Reasons for Negative Sentiment:")
print(neg_reason)

# Plot
plt.figure()
neg_reason.plot(kind='bar')
plt.title("Reasons for Negative Tweets")
plt.xlabel("Reason")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```
