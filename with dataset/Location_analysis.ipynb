```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset (update path if needed)
df = pd.read_csv("/content/Tweets.csv")

# Check columns
print(df.columns)

# Use correct column
location_col = "tweet_location"

# Step 1: Remove missing values
df_loc = df[df[location_col].notna()]

# Step 2: Clean locations (lowercase + strip spaces)
df_loc[location_col] = df_loc[location_col].str.strip().str.lower()

# Step 3: Get top 10 locations
top_locations = df_loc[location_col].value_counts().head(10)

print("\nTop 10 Locations:")
print(top_locations)

# Step 4: Visualization
plt.figure()
top_locations.plot(kind='bar')
plt.title("Top Tweet Locations")
plt.xlabel("Location")
plt.ylabel("Number of Tweets")
plt.xticks(rotation=45)
plt.show()
```
